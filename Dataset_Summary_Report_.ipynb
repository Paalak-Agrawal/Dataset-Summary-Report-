```python
# ============================================================
# ASSIGNMENT: DATASET SUMMARY REPORT (DAY 7)
# ============================================================

# Import Pandas library
import pandas as pd


# ============================================================
# 1. LOAD THE CSV DATASET
# ============================================================

# Replace "dataset.csv" with your actual CSV file name
df = pd.read_csv("dataset.csv")

print("Dataset loaded successfully!")


# ============================================================
# 2. DISPLAY THE FIRST 5 ROWS
# ============================================================

print("\n--- FIRST 5 ROWS OF THE DATASET ---")
print(df.head())


# ============================================================
# 3. DISPLAY THE LAST 5 ROWS
# ============================================================

print("\n--- LAST 5 ROWS OF THE DATASET ---")
print(df.tail())


# ============================================================
# 4. DISPLAY RANDOM SAMPLE ROWS
# ============================================================

print("\n--- RANDOM SAMPLE OF 5 ROWS ---")
print(df.sample(5))


# ============================================================
# 5. CHECK THE DATASET SHAPE
# ============================================================

print("\n--- DATASET SHAPE ---")

print("Dataset Shape:", df.shape)

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])


# ============================================================
# 6. DISPLAY COLUMN NAMES
# ============================================================

print("\n--- COLUMN NAMES ---")

print(df.columns)


# ============================================================
# 7. DISPLAY THE DATASET INDEX
# ============================================================

print("\n--- DATASET INDEX ---")

print(df.index)


# ============================================================
# 8. CHECK DATA TYPES
# ============================================================

print("\n--- DATA TYPES ---")

print(df.dtypes)


# ============================================================
# 9. DATASET INFORMATION
# ============================================================

print("\n--- DATASET INFORMATION ---")

df.info()


# ============================================================
# 10. DESCRIPTIVE STATISTICS
# ============================================================

print("\n--- DESCRIPTIVE STATISTICS ---")

print(df.describe())


# ============================================================
# 11. CHECK MISSING VALUES
# ============================================================

print("\n--- MISSING VALUES ---")

print(df.isnull().sum())


# ============================================================
# 12. TOTAL MISSING VALUES
# ============================================================

total_missing = df.isnull().sum().sum()

print("\nTotal Missing Values in Dataset:", total_missing)


# ============================================================
# 13. BASIC DATASET OBSERVATIONS
# ============================================================

print("\n--- BASIC DATASET OBSERVATIONS ---")

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nAvailable Features:")
for column in df.columns:
    print("-", column)

print("\nData Types:")
for column, data_type in df.dtypes.items():
    print(column, ":", data_type)

print("\nColumns with Missing Values:")

missing_values = df.isnull().sum()

found_missing = False

for column, count in missing_values.items():
    if count > 0:
        print(column, ":", count)
        found_missing = True

if found_missing == False:
    print("No missing values found.")


# ============================================================
# 14. CONCLUSION
# ============================================================

print("\n--- CONCLUSION ---")

print("The dataset contains", df.shape[0], "rows and", df.shape[1], "columns.")
print("The dataset includes the features listed above.")
print("The data types and missing values were examined.")
print("Descriptive statistics were generated for numerical columns.")
print("The dataset has been successfully explored using Pandas.")


# ============================================================
# END OF PROGRAM
# ============================================================

print("\nDataset Summary Report completed successfully!")
```
